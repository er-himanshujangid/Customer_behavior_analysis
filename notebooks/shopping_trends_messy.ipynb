```python
df["age"] = df["age"].fillna(df["age"].median()).astype(int)
df["review_rating"] = df["review_rating"].fillna(df["review_rating"].median())

for c in ["season", "size"]:
    df[c] = df[c].fillna(df[c].mode()[0])
```

```python
(df["discount_applied"] == df["promo_code_used"]).mean()   # 1.0
df = df.drop(columns=["promo_code_used"])
df.shape
```

```python
df["age_group"] = pd.cut(
    df["age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]
)
df["age_group"].value_counts().sort_index()
```

```python
df["spend_level"] = pd.cut(
    df["purchase_amount_usd"],
    bins=[0, 40, 70, 100],
    labels=["Low", "Medium", "High"]
)
df["spend_level"].value_counts().sort_index()
```
